# Arabian Nights Event Dataset Analysis

This notebook checks and summarizes our manually coded dataset. Put `jinn_folklore_60.csv` in the same folder as this notebook, then run all cells. No extra packages are needed.


```python
import csv
from collections import Counter
from pathlib import Path

DATA = Path('jinn_folklore_60.csv')
REQUIRED = [
    'record_id', 'story_title', 'motif_code', 'actor_type',
    'supernatural_role', 'jinn_explicit', 'source_url', 'source_evidence'
]
```

## Load the data and calculate totals


```python
with DATA.open(encoding='utf-8-sig', newline='') as file:
    reader = csv.DictReader(file)
    rows = list(reader)
    columns = reader.fieldnames or []

missing = [name for name in REQUIRED if name not in columns]
if missing:
    raise ValueError('Missing columns: ' + ', '.join(missing))

print('Total events:', len(rows))
print('Distinct story sections:', len(set(row['story_title'] for row in rows)))
print('Explicit jinn events:', sum(row['jinn_explicit'].lower() == 'yes' for row in rows))
print('Unique motif codes:', len(set(row['motif_code'] for row in rows)))
```

## Category counts


```python
for field in ['story_title', 'motif_code', 'supernatural_role', 'actor_type']:
    print('\n' + field.replace('_', ' ').title())
    for label, count in Counter(row[field] or '(blank)' for row in rows).most_common():
        print(f'  {label}: {count}')
```

## Validate the dataset


```python
problems = []
ids = [row['record_id'] for row in rows]
if len(ids) != len(set(ids)):
    problems.append('Duplicate record IDs')

for index, row in enumerate(rows, start=2):
    for field in REQUIRED:
        if not row[field].strip():
            problems.append(f'CSV line {index}: missing {field}')
    if row['jinn_explicit'].lower() not in ('yes', 'no'):
        problems.append(f'CSV line {index}: invalid jinn_explicit value')

print('\nBasic validation:', 'PASS' if not problems else 'REVIEW NEEDED')
for problem in problems:
    print(' -', problem)
print('\nNote: basic validation does not prove that source passages or interpretations are correct.')
```
